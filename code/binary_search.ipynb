```python
import random

list_numbers = list(range(0,200))
selected_number = random.randint(0,200)

def binary_search(list_numbers:list, selected_number:int):
    low = 0
    high = len(list_numbers) - 1

    while high >= low:
        mid = (high + low) // 2

        if list_numbers[mid] < selected_number:
            low = mid + 1
        elif list_numbers[mid] > selected_number:
            high = mid - 1
        
        else:
            return mid
    return -1


result = binary_search(list_numbers, selected_number)
 
if result != -1:
    print("Element is", str(selected_number))
    print("Element is present at index", str(result))
else:
    print("Element is not present in array")
```
